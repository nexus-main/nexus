Create client and authenticate

```typescript
import { NexusClient, V2 } from "npm:nexus-api"

// You get this token in the user settings menu of Nexus.
const accessToken = "<token>"
const baseUrl = "http://localhost:5000"
const client = new NexusClient(baseUrl)

client.signIn(accessToken)
```

Export data from sample catalog `/SAMPLE/LOCAL`

```typescript
const begin = "2020-01-01T00:00:00Z"
const end = "2020-01-02T00:00:00Z"

const T1 = "/SAMPLE/LOCAL/T1/1_s"
const V1 = "/SAMPLE/LOCAL/V1/1_s"

// Use a file period of PT0S to write all data into a single file.
const filePeriod = "PT0S"

// The following writers are currently supported:
// "Nexus.Writers.Csv" | "Nexus.Writers.Hdf5" | "Nexus.Writers.Mat73" | "Nexus.Writers.Famos"
const fileFormat = "Nexus.Writers.Csv"

// Nexus.Writers.Csv supports the following optional request configuration parameters:
// https://github.com/nexus-main/nexus/blob/master/src/Nexus/Extensions/Writers/README.md
const configuration: Record<string, unknown> = {
  "significant-figures": "4",
  "row-index-format": "iso-8601",
}

await client.export(
  begin,
  end,
  filePeriod,
  fileFormat,
  [T1, V1],
  configuration,
  "data",
  V2.Precision.Float64,
  (progress, message) => console.log(`${(progress * 100).toFixed(0).padStart(3)} %: ${message}`)
)
```
