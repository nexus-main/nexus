Create client and authenticate

```typescript
import { NexusClient, V2 } from "npm:nexus-api"

// You get this token in the user settings menu of Nexus.
const accessToken = "<token>"
const baseUrl = "http://localhost:5000"
const client = new NexusClient(baseUrl)

client.signIn(accessToken)
```

Load data from sample catalog `/SAMPLE/LOCAL`

```typescript
const begin = "2020-01-01T00:00:00Z"
const end = "2020-01-01T00:02:00Z"

const T1 = "/SAMPLE/LOCAL/T1/1_s"
const T1_MEAN = "/SAMPLE/LOCAL/T1/5_s_mean"

const data = await client.load(
  begin,
  end,
  [T1, T1_MEAN],
  V2.Precision.Float32,
  progress => console.log(`${(progress * 100).toFixed(0).padStart(3)} %: Loading`)
)

const t1 = data[T1]
const t1Mean = data[T1_MEAN]

console.log(`${T1}: ${t1.values.length} samples`)
console.log(`${T1_MEAN}: ${t1Mean.values.length} samples`)
```
